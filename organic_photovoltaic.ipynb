**The relation between properties**

$$
\begin{align}
gap = HOMO - LUMO
\end{align}
$$
with
* $HOMO$ = highest occcupied molecular orbital (eV)
* $LUMO$ = lowest unoccupied molecular orbital (eV)

and
$$
\begin{align}
PCE = 100 \times \frac{V_{oc} \times FF \times J_{sc}}{P_{in}}
\end{align}
$$
With
* $PCE$ = power conversion efficiency (%)
* $V_{oc}$ =  open circuit potential (V)
* $FF$ = fill factor
* $J_{sc}$ = short circuit density (mA cm⁻²)
* $P_{in}$ = input power

#### **3.2.3 Outliers**

````python
from scipy.stats import zscore

z_threshold = 3
def zscore_criteria(target):
  z_val = np.abs(zscore(df[target]))
  mask_clean = z_val < z_threshold
  result = df[~mask_clean][target]
  return [mask_clean, result]
````

````python
mask_clean, outliers = zscore_criteria("LUMO")
df = df[mask_clean].copy()
mask_clean, outliers = zscore_criteria("HOMO")
df = df[mask_clean].copy()
mask_clean, outliers = zscore_criteria("Voc")
df = df[mask_clean].copy()
````

````python
from rdkit import Chem
from rdkit.Chem import rdFingerprintGenerator
from rdkit.Chem import AllChem
import numpy as np

morgan_fp = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)
def smile_to_fp(x_smile):
  x_mol = Chem.MolFromSmiles(x_smile)
  x_fp = morgan_fp.GetFingerprint(x_mol)
  return list(x_fp)

# apply it to all the conor column
donor_fp = df["Donor"].apply(lambda x: smile_to_fp(x)).tolist()

# change to dataframe
X_donor = pd.DataFrame(donor_fp, columns=[f"fp_{i}" for i in range(2048)])
X_donor.head()
````

````python
# use pandas for one hot encoding
X_acceptor = pd.get_dummies(df["Acceptor"])
X_acceptor
````

````python
# concantenate the feature of donor and acceptor
X = pd.concat([X_donor.reset_index(drop=True), X_acceptor.reset_index(drop=True)], axis=1)
X_donor.shape, X_acceptor.shape, X.shape
````

````python
# select all target
target_cols = ["HOMO", "LUMO", "Gap", "PCE", "Voc", "Joc"]
y = df[target_cols].reset_index(drop=True)
y
````

#### **3.3.4 Train Test Split**

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=67)
````

#### **3.3.5 Scaling**

````python
from sklearn.preprocessing import StandardScaler

# scaling, make the mean=0
scaler_X = StandardScaler()

# only in train
scaler_X.fit(X_train)

# apply to train and test
X_train_scaled = scaler_X.transform(X_train)
X_test_scaled = scaler_X.transform(X_test)

# do the same thing with y
scaler_y = StandardScaler()
scaler_y.fit(y_train)
y_train_scaled = scaler_y.transform(y_train)
y_test_scaled = scaler_y.transform(y_test)
````

### **3.4 Model Development**

##### **3.4.1 Multi Layer Perceptron**

<div align="center">
  <img src="https://www.researchgate.net/publication/348236232/figure/fig1/AS:11431281246201285@1716319530636/Representation-of-multilayer-perceptron-MLP-with-three-hidden-layers-and-five-SNP.tif" width="75%" />
  <p><i>Figure 9: Representation of multilayer perceptron (MLP) with three hidden layers and five input</i></p>

````python
from sklearn.neural_network import MLPRegressor

# make a simple model
model = MLPRegressor(
    hidden_layer_sizes=(64, 32),
    activation="relu",
    solver="adam",
    max_iter=1000,
    random_state=67
)

# train the model
model.fit(X_train_scaled, y_train_scaled)
````

##### **3.4.2 Prediction**

````python
# predict
y_pred_scaled = model.predict(X_test_scaled)

# inverse transform
y_pred = scaler_y.inverse_transform(y_pred_scaled)
````

##### **3.4.3 Model Evaluation**

Since this is a regression problem, the evaluation metrics that we use are MSE (mean squared error), RMSE (root mean squared error), and R² (coefficient of determination).

$$
\begin{align}
MSE &= \sum_{i=1}^N(y_i - y_i')^2 \\
RMSE &= \sqrt{\sum_{i=1}^N(y_i - y_i')^2} \\
R^2 &= 1 - \frac{MSE}{Var(y)}
\end{align}
$$

* $N$ = number of data
* $y_i$ = target variable
* $y_i'$ = model prediction
* $Var(y)$ = the variance of sample data


````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# evaluation metrics
metrics_list = []

# target_cols = ["HOMO", "LUMO", "Gap", "PCE", "Voc", "Joc"]
for i, col_name in enumerate(target_cols):
  # take data each column
  actual = y_test.values[:, i]
  pred = y_pred[:, i]

  # metrics evaluation
  mse = mean_squared_error(actual, pred)
  rmse = np.sqrt(mse)
  r2 = r2_score(actual, pred)

  # save
  metrics_list.append({
      "Property": col_name,
      "R2 Score": round(r2, 2),
      "RMSE": round(rmse, 3)
  })

evaluation_matrix = pd.DataFrame(metrics_list)
evaluation_matrix
````

````python
from sklearn.model_selection import GridSearchCV
from sklearn.neural_network import MLPRegressor
from sklearn.model_selection import RandomizedSearchCV

# Define the expanded hyperparameter search space
param_grid_expanded = {
"hidden_layer_sizes": [
        (64,), (128,), (256,),
        (64, 32), (128, 64), (256, 128),
        (128, 128), (256, 256),
        (128, 64, 32), (256, 128, 64)
    ],
    "activation": ["relu", "tanh", "identity", "logistic"],
    "alpha": [0.00001, 0.0001, 0.001, 0.01, 0.1],
    "learning_rate_init": [0.0001, 0.0005, 0.001, 0.005, 0.01, 0.05, 0.1],
    "solver": ["adam", "sgd"],
    "learning_rate": ["constant", "adaptive"]
}

# Initialize the MLPRegressor model
mlp = MLPRegressor(max_iter=1000, random_state=67)

grid_search = RandomizedSearchCV(
    estimator=mlp,           # Ganti dengan variabel model MLP Anda
    param_distributions=param_grid_expanded,
    n_iter=200,               # Jumlah eksperimen yang dijalankan
    cv=5,                    # Cross-validation 5-fold
    verbose=2,
    random_state=67,
    n_jobs=1
)

grid_search.fit(X_train_scaled, y_train_scaled)

print("Grid search is finished!\n")
print(f"Best parameters: {grid_search.best_params_}")
````

````python
best_model = grid_search.best_estimator_

y_pred_scaled = best_model.predict(X_test_scaled)
y_pred = scaler_y.inverse_transform(y_pred_scaled)

# evaluation metrics
metrics_list = []

# target_cols = ["HOMO", "LUMO", "Gap", "PCE", "Voc", "Joc"]
for i, col_name in enumerate(target_cols):
  # take data each column
  actual = y_test.values[:, i]
  pred = y_pred[:, i]

  # metrics evaluation
  mse = mean_squared_error(actual, pred)
  rmse = np.sqrt(mse)
  r2 = r2_score(actual, pred)

  # save
  metrics_list.append({
      "Property": col_name,
      "R2 Score": round(r2, 2),
      "RMSE": round(rmse, 3)
  })

evaluation_matrix = pd.DataFrame(metrics_list)
print(evaluation_matrix)
````

## **4. Problems and Future Prospetive**



### **4.1 Data Infrastructure**
* Harvard Clean Energi Project used, but usually real materials are more complicated than that.
* Limited data of OPV (compare to other ML problem like image recognition) and costly DFT calculation.
* Solution: bayesian framwork, precision and DoF trade off.


### **4.2 Descriptor Selection**
* Molecular descriptors are fast and easy to calculate but not completely suitable.
* Microscopic properties that related to photovoltaic phenomena are required costly computations (bandgap `->` open circuit voltage; effective mass `->` charge mobility, short circuit current; reorganization energy `->` fill factor).
* Solution: need accurate, easily accessible, and new generation of descriptors specifically designed for OPV.


### **4.3 Multidimension Design**

Add additional data :
* Transmission electron microscopy (TEM)
* Atomic force microscopy (AFM)
* Grazing-incidence small-angle x-ray scattering (GISAXS)
*  Grazing incidence wide angle x-ray scattering (GIWAXS)




### **4.4 Experimental Validation**
* No guarantee we can synthesis the results.
* We don’t know the procedures.
* Solution: Need collaboration with experimental expert.



### **4.5 Developement of Better Software**
* Many experimentalist don't have background in ML.
* Only they who know programming and data science can do it.
* They don’t know the underlying process (it can lead to wrong interpretation of results).
* Solution: Make user friendly ML software, LLM.
