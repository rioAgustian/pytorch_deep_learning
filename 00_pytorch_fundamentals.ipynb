### Matrix multiplication

Two main ways of performing multiplication in neural networks and deep learning:

1. Element-wise multiplication
2. Matrix multiplication (dot product)

There are two main rules that performing matrix multiplication needs to satify:
1. The **inner dimensions** must match:
* `(3, 2) @ (3, 2)` won't work
* `(2, 3) @ (3, 2)` will work
* `(3, 2) @ (2, 3)` will work
2. The resulting matrix has the shape of the **outter dimension**:
* `(2, 3) @ (3, 2)` -> `(2, 2)`
* `(3, 2) @ (2, 3)` -> `(3, 3)`

````python
torch.matmul(torch.rand(10, 10), torch.rand(10, 10)).shape
````

### One of the most common errors in dep learning: shape errors

````python
# shapes for matrix multiplication
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])
tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])
torch.matmul(tensor_A, tensor_B)
````

````python
tensor_A.shape, tensor_B.shape
````

To fix our tensor shape issue, we can manipulate the shape of one of our tensor using a **transpose**.

A **transpose** switches the axes or dimension of given tensor.

````python
tensor_B, tensor_B.shape
````

````python
tensor_B.T, tensor_B.T.shape
````

````python
# The matrix multiplication operation works when tensor_B is transposed
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"New shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.T.shape}")
print(f"Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <- inner dimension must match")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"\nOutput shape: {output.shape}")
````

## Reshaping, stacking, squeezing, and unsqueezing

* Reshaping -reshapes an input tensor to a defined shape.
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor.
* Stacking - Combine multiple tensor on top of each other (vstack) or side by side (hstack).
* Squeeze - Removes all `1` dimensions from a tensor.
* Unsqueeze - Adds a `1` dimension to a target tensor.
* Permute - Return a view of the input with dimension permuted (sqapped) in a certain way.

````python
# Let's create a tensor
import torch

x = torch.arange(1., 10.)
x, x.shape
````

````python
# Add an extra dimension
x_reshape = x.reshape(1, 9)
x_reshape, x_reshape.shape
````

````python
# Change the view
z = x.view(1, 9)
z, z.shape
````

````python
# Changin z changes x (because a view of atensor share the same memory as the original input)
z[:, 0] = 5
z, x
````

````python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x])
x_stacked
````

````python
# torch.squeez() - removes all single dimension from a target tensor
print(f"Previous tensor: {x_reshape}")
print(f"Previous shape: {x_reshape.shape}")

# Remove extra dimensions from x_reshape
x_squeeze = x_reshape.squeeze()
print(f"\nNew tensor: {x_squeeze}")
print(f"New shape: {x_squeeze.shape}")
````

````python
# torch.unsqueeze() - adds a single dimension to a target tensor at a specific dimension
print(f"Previous target: {x_squeeze}")
print(f"Previous shape: {x_squeeze.shape}")

# Add an extra dimension with unsqueeze
x_unsqueeze = x_squeeze.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueeze}")
print(f"New shape: {x_unsqueeze.shape}")
````

````python
# torch.permute() - rearranges the dimensions of a target tensor in a specifiez order

x_permute = torch.randn(2, 3, 5)
x_permute.size()
````

````python
torch.permute(x_permute, (2, 0, 1)).size() # dimension 2 become 0, 0 become 1, 1 become 2
````

````python
x_original = torch.rand(size=(224, 224, 3)) # [height, widht, color_channel]

# Permute the original tensor to rearrange the axis (or dim) order
x_permute = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"Previous tensor shape: {x_original.shape}")
print(f"New tensor shape: {x_permute.shape}")
````

````python
x_original[0, 0, 0] = 9999
x_original[0, 0, 0], x_permute[0, 0, 0]
````

## PyTorch tensors & NumPy

Numpy is a popular scientific Python numerical computing library.

And because of this, PyTorch has functionality to interact with it.

* Data in NumPy, want in PyTorch tensor -> `torch.from_numpy(ndarray)`
* PyTorch tensor -> NumPy -> `torch.Tensor.numpy()`

````python
# Numpy array to tensor
import torch
import numpy as np

array = np.arange(1., 8.)
tensor = torch.from_numpy(array) # you need to convert to float32
array, tensor
````

## Rreproducability (trying to take random out of random)

In short how a neural network learns:

`Starts with random numbers -> tensor operations -> update random numbers to try and make them better representation of the data -? again -> again -> again -> ...`

To reduce the randomness in neural netweorks and PyTorch comes the concept of a **random seed**.

Esentially what the random seed does is "flavour" the randomness.

````python
import torch

torch.manual_seed(100)
# Create two random tensors
random_tensor_A = torch.rand(3, 4)

torch.manual_seed(100)
random_tensor_B = torch.rand(3, 4)
print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B
      )
````

````python
# Let's make random but reproducable tensors
import torch

# Set the random seed
random_seed = 100

torch.manual_seed(random_seed)
random_tensor_c = torch.rand(3, 4)

torch.manual_seed(random_seed)
random_tensor_d = torch.rand(3, 4)

print(random_tensor_c)
print(random_tensor_d)
print(random_tensor_c==random_tensor_d)
````
