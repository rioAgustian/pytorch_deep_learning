## Exercise 4

```python
import torch

seed = 0
torch.manual_seed(seed)
tensor_A = torch.rand(7, 7)
print(f"Tensor A = {tensor_A}")

torch.manual_seed(seed)
tensor_B = torch.rand(1, 7)
torch.manual_seed(seed)
tensor_C = torch.rand(1, 7)
tensor_C = tensor_C.T
result = torch.matmul(tensor_B, tensor_C)
print(f"\nTensor B = {tensor_B}")
print(f"Tensor C = {tensor_C}")
print(f"\nMatrix multiplication of tensor B and tensor C = {result}")
```

## Exercise 6

```python
import torch

seed = 1234
torch.manual_seed(seed)
tensor_A = torch.randn(2, 3, device='cuda')
torch.manual_seed(seed)
tensor_B = torch.randn(2, 3, device='cuda')
```

## Exercise 7

```python
result = torch.matmul(tensor_A, tensor_B.T)
print(f"Tensor A = {tensor_A}")
print(f"Tensor B = {tensor_B.T}")
print(f"\nTensor A @ tensor B = {result}")
```

## Exercise 8

```python
max = result.max()
min = result.min()
print(f"Min = {min}")
print(f"Max = {max}")
```

## Exercise 9

```python
ind_max = result.argmax()
ind_min = result.argmin()
print(f"Index of min = {ind_min}")
print(f"Index of max = {ind_max}")
```
